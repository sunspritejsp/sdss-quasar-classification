```python
df.groupby('class')['redshift'].describe().round(4)
```

```python
not_quasars = ((df['class'] == 'QSO') & (df['redshift'] < 0.05))
print(not_quasars.sum())
print(df[not_quasars].head())
```

```python
df[df['redshift'] > 0.1].head()
```

```python
star_anomalies = ((df['class'] == 'STAR') & ((df['redshift'] < -0.005) | (df['redshift'] > 0.002)))
print('number of star anomalies: ', star_anomalies.sum())
print('duplicate objIDs: ', df['objID'].duplicated().sum())
```

confirms `zWarning` filter works. no abnormal stars found, no duplicated objects too.

```python
bad_sentinels = (df[bands] == -9999).any(axis=1)
bad_faint = (df[bands] > 30).any(axis=1)
bad_qso_z = (df['class'] == 'QSO') & (df['redshift'] < 0.05)

bad_rows = bad_faint | bad_qso_z | bad_sentinels

df_clean = df[~bad_rows].copy()

print('before: ', len(df))
print('after cleaning: ', len(df_clean))
print('diff: ', bad_rows.sum())
```

```python
df_clean.to_csv("sdss_clean.csv", index=False)
```
